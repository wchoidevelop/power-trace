```python
with open('lu.json', 'r', encoding='utf-8') as f:
    lu = json.load(f)

with open('le-indy.json', 'r', encoding='utf-8') as f:
    indy = json.load(f)

with open('le-occupation.json', 'r', encoding='utf-8') as f:
    occupation = json.load(f)

#templates
#union representation percentage
union_represented = {
    "lfst_code": "20",
    "fips_code": "00",
    'tdata_code': '01',
    'pcts_code': '05',
    "earn_code": "00",
    'class_code': '17',
    "unin_code": "2",
    "education_code": "00",
    "ages_code": "00",
    "race_code": "00",
    "orig_code": "00",
    "sexs_code": "0",
    "seasonal": "U",
}

ur_o = dict(union_represented, indy_code='0000', class_code='16')
ur_i = dict(union_represented, occupation_code='0000')

union_represented_wage = {
    "lfst_code": "25",
    "fips_code": "00",
    'tdata_code': '00',
    'pcts_code': '00',
    "earn_code": "01",
    'class_code': '17',
    "unin_code": "2",
    "education_code": "00",
    "ages_code": "00",
    "race_code": "00",
    "orig_code": "00",
    "sexs_code": "0",
    "seasonal": "U",
}

urw_o = dict(union_represented_wage, indy_code='0000', class_code='16')
urw_i = dict(union_represented_wage, occupation_code='0000')

nonunion_wage = {
    "lfst_code": "25",
    "fips_code": "00",
    'tdata_code': '00',
    'pcts_code': '00',
    "earn_code": "01",
    'class_code': '17',
    "unin_code": "3",
    "education_code": "00",
    "ages_code": "00",
    "race_code": "00",
    "orig_code": "00",
    "sexs_code": "0",
    "seasonal": "U",
}

urw_o = dict(union_represented_wage, indy_code='0000', class_code='16')
urw_i = dict(union_represented_wage, occupation_code='0000')
nu_o = dict(nonunion_wage, indy_code='0000', class_code='16')
nu_i = dict(nonunion_wage, indy_code='0000', class_code='16')

coverage = {'occupation': {}, 'industry': {}}
wages = {'occupation': {}, 'industry': {}}
nonunion_wages = {'occupation': {}, 'industry': {}}

for row in lu:
    if ur_o.items() <= row.items():
        if coverage["occupation"].get(row['occupation_code']) is not None:
            print(f"duplicate for occupation coverage {row['occupation_code']}")
        coverage["occupation"][row['occupation_code']] = row
    elif ur_i.items() <= row.items():
        if coverage["industry"].get(row['indy_code']) is not None:
            print(f"duplicate for industry coverage {row['indy_code']}")
        coverage["industry"][row['indy_code']] = row
    elif urw_o.items() <= row.items():
        if wages["occupation"].get(row['occupation_code']) is not None:
            print(f"duplicate for occupation wages {row['occupation_code']}")
        wages["occupation"][row['occupation_code']] = row
    elif urw_i.items() <= row.items():
        if wages["industry"].get(row['indy_code']) is not None:
            print(f"duplicate for industry wages {row['indy_code']}")
        wages["industry"][row['indy_code']] = row
    elif nu_o.items() <= row.items():
        if nonunion_wages["occupation"].get(row['occupation_code']) is not None:
            print(f"duplicate for occupation wages {row['occupation_code']}")
        nonunion_wages["occupation"][row['occupation_code']] = row
    elif nu_i.items() <= row.items():
        if nonunion_wages["industry"].get(row['indy_code']) is not None:
            print(f"duplicate for industry wages {row['indy_code']}")
        nonunion_wages["industry"][row['indy_code']] = row

with open('lu_coverage_series_index.json', 'w', encoding='utf-8') as f:
    json.dump(coverage, f, ensure_ascii=False, indent=2)

with open('lu_wages_series_index.json', 'w', encoding='utf-8') as f:
    json.dump(wages, f, ensure_ascii=False, indent=2)
with open('lu_nonunion_wages_series_index.json', 'w', encoding='utf-8') as f:
    json.dump(nonunion_wages, f, ensure_ascii=False, indent=2)
```
